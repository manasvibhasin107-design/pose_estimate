```python

import cv2
import numpy as np
from ultralytics import YOLO

model = YOLO("yolov8n-pose.pt")
cap = cv2.VideoCapture("vimg.mp4")

w = int(cap.get(3))
h = int(cap.get(4))
fps = cap.get(5)

if fps <= 0:
    fps = 30

w, h = w - w % 2, h - h % 2

out = cv2.VideoWriter(
    "final_pose.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (w, h)
)

print("Writer opened:", out.isOpened())

skeleton = [(5,6),(5,7),(7,9),(6,8),(8,10),(5,11),
            (6,12),(11,12),(11,13),(13,15),
            (12,14),(14,16),(0,1),(0,2),(1,3),(2,4)]

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.resize(frame, (w, h))
    result = model(frame, verbose=False)[0]
    black = np.zeros_like(frame)

    for person in result.keypoints:
        points = person.xy[0].cpu().numpy()

        for a, b in skeleton:
            cv2.line(black, tuple(points[a].astype(int)),
                     tuple(points[b].astype(int)), (255,255,255), 2)

        for x, y in points.astype(int):
            cv2.circle(black, (x,y), 4, (255,255,255), -1)

    out.write(black)

cap.release()
out.release()
cv2.destroyAllWindows()

print("completed!")
```
